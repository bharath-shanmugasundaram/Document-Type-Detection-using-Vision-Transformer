```python
import os
import random
import shutil
from tqdm import tqdm

# ---------------- CONFIG ----------------
SOURCE_DIR = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Dataset"          # original dataset
TRAIN_DIR = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Dataset_root/Train"
DEV_DIR = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Dataset_root/Val"

TRAIN_COUNT = 8000
DEV_COUNT = 2000

IMAGE_EXTS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
SEED = 42  # for reproducibility
# ----------------------------------------

random.seed(SEED)

os.makedirs(TRAIN_DIR, exist_ok=True)
os.makedirs(DEV_DIR, exist_ok=True)

# Loop through each class folder
for class_name in os.listdir(SOURCE_DIR):
    class_path = os.path.join(SOURCE_DIR, class_name)

    if not os.path.isdir(class_path):
        continue

    print(f"\n📂 Processing class: {class_name}")

    # Create class folders in train/dev
    train_class_dir = os.path.join(TRAIN_DIR, class_name)
    dev_class_dir = os.path.join(DEV_DIR, class_name)

    os.makedirs(train_class_dir, exist_ok=True)
    os.makedirs(dev_class_dir, exist_ok=True)

    # Get all images
    images = [
        f for f in os.listdir(class_path)
        if f.lower().endswith(IMAGE_EXTS)
    ]

    if len(images) < (TRAIN_COUNT + DEV_COUNT):
        raise ValueError(
            f"Not enough images in {class_name}: found {len(images)}"
        )

    random.shuffle(images)

    train_images = images[:TRAIN_COUNT]
    dev_images = images[TRAIN_COUNT:TRAIN_COUNT + DEV_COUNT]

    # Copy train images
    for img in tqdm(train_images, desc=f"Train {class_name}"):
        src = os.path.join(class_path, img)
        dst = os.path.join(train_class_dir, img)
        shutil.copy2(src, dst)

    # Copy dev images
    for img in tqdm(dev_images, desc=f"Dev {class_name}"):
        src = os.path.join(class_path, img)
        dst = os.path.join(dev_class_dir, img)
        shutil.copy2(src, dst)

print("\n✅ Train / Dev split completed successfully!")
```
