```python
from datasets import load_dataset
import os
from tqdm import tqdm
from PIL import Image
import io

# Load dataset
ds = load_dataset("tusharshah2006/bank_statements_transactions")
# Inspect columns
print("Columns:", ds["train"].column_names)

# Change this if needed after seeing print
IMAGE_COLUMN = ds["train"].column_names[0]  # usually the image-like column

# Output directory
output_dir = "./invoices_images/train"
os.makedirs(output_dir, exist_ok=True)

# Save images
for idx, sample in enumerate(tqdm(ds["train"], desc="Saving images")):
    img_data = sample[IMAGE_COLUMN]

    # Case 1: already PIL Image
    if isinstance(img_data, Image.Image):
        img = img_data

    # Case 2: bytes (very common for OCR datasets)
    elif isinstance(img_data, (bytes, bytearray)):
        img = Image.open(io.BytesIO(img_data))

    # Case 3: dict with bytes
    elif isinstance(img_data, dict) and "bytes" in img_data:
        img = Image.open(io.BytesIO(img_data["bytes"]))

    else:
        raise TypeError(f"Unsupported image format: {type(img_data)}")

    img.save(f"{output_dir}/invoice_{idx}.png")

print(f"Saved {len(ds['train'])} images to {output_dir}")
```

```python
import os
import random
import shutil

SOURCE_ROOT = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Data Cleaner/filtered_images"
TARGET_ROOT = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/data_5class"

SAMPLES_PER_CLASS = 19900
SEED = 42
IMG_EXTENSIONS = (".jpg", ".jpeg", ".png")

CLASS_MAPPING = {
    "Invoice": ["invoice"],
    "Resume": ["resume"],
    "Government_Form": ["form", "questionnaire"],
    "Bank_Statement": ["budget"],
    "Insurance_Policy": ["letter", "memo", "specification", "scientific report"]
}

random.seed(SEED)
os.makedirs(TARGET_ROOT, exist_ok=True)

def get_images(folder):
    return [
        os.path.join(folder, f)
        for f in os.listdir(folder)
        if f.lower().endswith(IMG_EXTENSIONS)
    ]

for target_class, source_classes in CLASS_MAPPING.items():
    print(f"\nProcessing {target_class}")
    all_images = []

    # 🔴 CLEAR TARGET FOLDER FIRST
    target_dir = os.path.join(TARGET_ROOT, target_class)
    if os.path.exists(target_dir):
        shutil.rmtree(target_dir)
    os.makedirs(target_dir, exist_ok=True)

    for src in source_classes:
        src_path = os.path.join(SOURCE_ROOT, src)

        if not os.path.exists(src_path):
            print(f"⚠️ Folder not found: {src_path}")
            continue

        all_images.extend(get_images(src_path))

    print(f"Found {len(all_images)} images")

    if len(all_images) < SAMPLES_PER_CLASS:
        raise ValueError(
            f"Not enough images for {target_class}: "
            f"{len(all_images)} < {SAMPLES_PER_CLASS}"
        )

    selected_imgs = random.sample(all_images, SAMPLES_PER_CLASS)

    for img in selected_imgs:
        shutil.copy(img, target_dir)

    print(f"✅ Copied EXACTLY {SAMPLES_PER_CLASS} images → {target_dir}")
```

```python
import cv2
import os
import shutil

# -------- CONFIG --------
SOURCE_ROOT = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Dataset"
TARGET_ROOT = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Dataset9"
IMG_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp")
WINDOW_NAME = "Review | ENTER=accept | SPACE=skip | ESC=exit"
# ------------------------

os.makedirs(TARGET_ROOT, exist_ok=True)

class_names = sorted([
    d for d in os.listdir(SOURCE_ROOT)
    if os.path.isdir(os.path.join(SOURCE_ROOT, d))
])

print(f"Found classes: {class_names}")
req = {"Government_Form", "Insurance_Policy"}
for class_name in class_names:
    if class_name not in req:
        continue
    src_class_dir = os.path.join(SOURCE_ROOT, class_name)
    tgt_class_dir = os.path.join(TARGET_ROOT, class_name)

    os.makedirs(tgt_class_dir, exist_ok=True)

    images = [
        f for f in sorted(os.listdir(src_class_dir))
        if f.lower().endswith(IMG_EXTENSIONS)
    ]

    print(f"\n📂 Class: {class_name} | Images: {len(images)}")

    for idx, img_name in enumerate(images, 1):
        img_path = os.path.join(src_class_dir, img_name)

        img = cv2.imread(img_path)
        if img is None:
            print(f"[SKIP] Cannot read {img_name}")
            continue

        display = img.copy()
        cv2.putText(
            display,
            f"{class_name}  {idx}/{len(images)}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

        cv2.imshow(WINDOW_NAME, display)

        while True:
            key = cv2.waitKey(0)

            # ENTER → accept
            if key == 13:
                shutil.copy2(
                    img_path,
                    os.path.join(tgt_class_dir, img_name)
                )
                print(f"[ACCEPTED] {class_name}/{img_name}")
                break

            # SPACE → reject
            elif key == 32:
                print(f"[REJECTED] {class_name}/{img_name}")
                break

            # ESC → exit
            elif key == 27:
                print("❌ Exiting review")
                cv2.destroyAllWindows()
                exit()

cv2.destroyAllWindows()
print("✅ Dataset review completed")
```

```python
import os
import random
import shutil

# -------- CONFIG --------
SOURCE_ROOT = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Dataset"
TARGET_ROOT = "/Users/bhara-zstch1566/Projects/CRM DI/Document Type Detection/Dataset9"
TARGET_COUNT = 10000
IMG_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp")
SEED = 42
# ------------------------

random.seed(SEED)

for class_name in os.listdir(TARGET_ROOT):
    tgt_class_dir = os.path.join(TARGET_ROOT, class_name)
    src_class_dir = os.path.join(SOURCE_ROOT, class_name)

    if not os.path.isdir(tgt_class_dir):
        continue

    print(f"\n📂 Processing class: {class_name}")

    tgt_images = [
        f for f in os.listdir(tgt_class_dir)
        if f.lower().endswith(IMG_EXTENSIONS)
    ]

    # ---------------- TRIM IF EXCESS ----------------
    if len(tgt_images) > TARGET_COUNT:
        keep = set(random.sample(tgt_images, TARGET_COUNT))
        remove = set(tgt_images) - keep

        for img in remove:
            os.remove(os.path.join(tgt_class_dir, img))

        print(f"✂ Trimmed from {len(tgt_images)} → {TARGET_COUNT}")

    # ---------------- ADD IF SHORT ----------------
    elif len(tgt_images) < TARGET_COUNT:
        needed = TARGET_COUNT - len(tgt_images)

        if not os.path.isdir(src_class_dir):
            print(f"⚠ No source folder for {class_name}")
            continue

        src_images = [
            f for f in os.listdir(src_class_dir)
            if f.lower().endswith(IMG_EXTENSIONS)
        ]

        available = list(set(src_images) - set(tgt_images))

        if not available:
            print("⚠ No new images available to add")
            continue

        add_count = min(needed, len(available))
        selected = random.sample(available, add_count)

        for img in selected:
            shutil.copy2(
                os.path.join(src_class_dir, img),
                os.path.join(tgt_class_dir, img)
            )

        print(f"➕ Added {add_count} images ({len(tgt_images)+add_count}/10k)")

    else:
        print("✅ Already balanced (10k)")
```
